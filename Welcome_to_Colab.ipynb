```
# Question 1: Supermarket Bill Generator

apple_price = float(input("Enter apple price: "))
apple_qty = int(input("Enter apple quantity: "))

banana_price = float(input("Enter banana price: "))
banana_qty = int(input("Enter banana quantity: "))

apple_total = apple_price * apple_qty
banana_total = banana_price * banana_qty

total = apple_total + banana_total

print("\n----- BILL -----")
print("Apple Total:", apple_total)
print("Banana Total:", banana_total)
print("Total Bill:", total)
```

```
# Question 10: Pass / Fail Evaluation

m1 = float(input("Enter first subject marks: "))
m2 = float(input("Enter second subject marks: "))
m3 = float(input("Enter third subject marks: "))

average = (m1 + m2 + m3) / 3

if m1 < 33 or m2 < 33 or m3 < 33:
    print("FAIL: Marks below 33")
elif average >= 40:
    print("PASS")
else:
    print("FAIL: Average below 40")
```

```
# Question 12: Shopping Discount

bill = float(input("Enter bill amount: "))

if bill > 100:
    discount_percent = 20
elif bill >= 50:
    discount_percent = 10
else:
    discount_percent = 0

discount = bill * discount_percent / 100
final_bill = bill - discount

print("Discount Percentage:", discount_percent, "%")
print("Discount Amount:", discount)
print("Final Bill:", final_bill)
```

```
# Question 13: Electricity Bill

units = float(input("Enter electricity units: "))

if units <= 100:
    bill = units * 0.50
elif units <= 200:
    bill = 100 * 0.50 + (units - 100) * 0.75
else:
    bill = 100 * 0.50 + 100 * 0.75 + (units - 200) * 1.20

print("Electricity Bill: $", round(bill, 2))
```

```
# Question 14: ATM Withdrawal

balance = float(input("Enter account balance: "))
amount = int(input("Enter withdrawal amount: "))

if amount <= 0:
    print("Invalid withdrawal amount")
elif amount > balance:
    print("Insufficient Balance")
elif amount % 100 != 0:
    print("Amount must be a multiple of 100")
else:
    balance = balance - amount
    print("Withdrawal Successful")
    print("Remaining Balance:", balance)
```

```
# Question 15: Leap Year Checker

year = int(input("Enter year: "))

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

```
# Question 16: Triangle Type

a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a <= 0 or b <= 0 or c <= 0 or a + b <= c or a + c <= b or b + c <= a:
    print("Invalid Triangle")
elif a == b and b == c:
    print("Equilateral Triangle")
elif a == b or b == c or a == c:
    print("Isosceles Triangle")
else:
    print("Scalene Triangle")
```

```
# Question 20: PIN Verification

correct_pin = "1234"
attempts = 0

while attempts < 3:
    pin = input("Enter your 4-digit PIN: ")
    attempts += 1

    if pin == correct_pin:
        print("PIN Correct. Access Granted")
        break
    else:
        print("Incorrect PIN")

if attempts == 3 and pin != correct_pin:
    print("Account Locked")
```

```
# Question 21: Inventory Countdown

stock = int(input("Enter initial stock: "))

while stock > 0:
    stock = stock - 1
    print("Remaining Stock:", stock)

print("Stock Finished")
```

```
# Question 23: Odd-Even Counter

even = 0
odd = 0

for i in range(1, 21):
    if i % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even Numbers Count:", even)
print("Odd Numbers Count:", odd)
```

```
# Question 24: Fuel Cost Calculator

def calculate_fuel_cost(distance, mileage, price_per_liter):
    fuel = distance / mileage
    cost = fuel * price_per_liter
    return cost

result = calculate_fuel_cost(100, 20, 100)

print("Total Fuel Cost:", result)
```

```
# Question 25: BMI Calculator

def get_bmi(weight, height):
    return weight / (height ** 2)

bmi = get_bmi(60, 1.7)

print("BMI:", round(bmi, 2))
```

```
# Question 26: Simple Interest

def calculate_simple_interest(principal, rate, time):
    interest = (principal * rate * time) / 100
    return interest

result = calculate_simple_interest(10000, 5, 2)

print("Simple Interest:", result)
```

```
# Question 27: Maximum of Three Numbers

def find_highest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

result = find_highest(10, 25, 15)

print("Highest Number:", result)
```

```
# Question 28: Currency Formatting

def format_currency(amount):
    return "$" + format(amount, ".2f")

result = format_currency(125.5)

print("Formatted Amount:", result)
```

```
# Question 29: Circle Details

def circle_details(radius):
    area = 3.14159 * radius ** 2
    circumference = 2 * 3.14159 * radius
    return area, circumference

area, circumference = circle_details(5)

print("Area:", round(area, 2))
print("Circumference:", round(circumference, 2))
```

```
# Question 30: Student Grade Classifier

def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 50:
        return "C"
    else:
        return "F"

score = float(input("Enter student percentage: "))

grade = get_grade(score)

print("Student Grade:", grade)
```
