```python
# WAP For Prime Number

n = int(input("Enter a number: "))

count = 0

for i in range(1, n + 1):
    if n % i == 0:
        count += 1

if count == 2:
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
# WAP For Prime Number Series
n = int(input("Enter the limit: "))

for num in range(2, n + 1):
    count = 0

    for i in range(1, num + 1):
        if num % i == 0:
            count += 1

    if count == 2:
        print(num, end=" ")
```

```python
# WAP For Armstrong Number

n = int(input("Enter a number: "))

temp = n
power = len(str(n))
total = 0

while temp > 0:
    digit = temp % 10
    total += digit ** power
    temp //= 10

if total == n:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
# WAP For Armstrong Series

n = int(input("Enter the limit: "))

for num in range(1, n + 1):
    temp = num
    power = len(str(num))
    total = 0

    while temp > 0:
        digit = temp % 10
        total += digit ** power
        temp //= 10

    if total == num:
        print(num, end=" ")
```

```python
# WAP For Perfect Number

n = int(input("Enter a number: "))

total = 0

for i in range(1, n):
    if n % i == 0:
        total += i

if total == n and n > 0:
    print("Perfect Number")
else:
    print("Not a Perfect Number")
```

```python
# WAP For Perfect Number Series

n = int(input("Enter the limit: "))

for num in range(1, n + 1):
    total = 0

    for i in range(1, num):
        if num % i == 0:
            total += i

    if total == num:
        print(num, end=" ")
```

```python
# WAP For Fibonacci Series

n = int(input("How many terms? "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
# WAP For Palindrome Number

n = int(input("Enter a number: "))

temp = n
reverse = 0

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp //= 10

if reverse == n:
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")
```

```python
# WAP For Palindrome Series

n = int(input("Enter the limit: "))

for num in range(1, n + 1):
    temp = num
    reverse = 0

    while temp > 0:
        digit = temp % 10
        reverse = reverse * 10 + digit
        temp //= 10

    if reverse == num:
        print(num, end=" ")
```
